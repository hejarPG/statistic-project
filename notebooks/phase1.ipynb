# ۲. واحد مشاهده و ساختار داده

واحد مشاهده، هر شرکت‌کننده NHANES است. در هر یک از چهار فایل پروژه، هر ردیف اطلاعات یک فرد را نشان می‌دهد و متغیر SEQN شناسه اوست. یکتا بودن این شناسه در کد زیر بررسی می‌شود.

یک فرد می‌تواند در چند فایل حضور داشته باشد؛ مثلا اطلاعات خواب او در SLQ_L و دور کمر او در BMX_L ثبت شده است. این‌ها اندازه‌گیری تکراری در زمان‌های مختلف نیستند، بلکه بخش‌های مختلف اطلاعات همان فرد هستند. پس از ادغام با SEQN، هر فرد فقط یک ردیف دارد و هر ستون یک متغیر را نشان می‌دهد.

کد زیر تعداد ردیف‌ها، شناسه‌های یکتا، تکراری و گمشده را محاسبه می‌کند. تعداد افراد مشترک چهار فایل نیز پس از ادغام به دست می‌آید.

</div>


```python
from pathlib import Path

project_root = next(
    (candidate for candidate in [Path.cwd(), *Path.cwd().parents]
     if (candidate / "PROJECT_CARD.md").is_file() and (candidate / "data").is_dir()),
    None
)
if project_root is None:
    raise FileNotFoundError("Open the notebook from inside the project directory.")

import pandas as pd

files = {
    name: pd.read_sas(project_root / "data" / "raw" / f"{name}.xpt", format="xport", encoding="utf-8")
    for name in ["DEMO_L", "SLQ_L", "BMX_L", "DPQ_L"]
}

structure_rows = [
    {
        "داده": name,
        "تعداد ردیف": len(df),
        "تعداد افراد یکتا": df["SEQN"].nunique(),
        "شناسه تکراری": int(df["SEQN"].duplicated().sum()),
        "شناسه گمشده": int(df["SEQN"].isna().sum())
    }
    for name, df in files.items()
]

for name, df in files.items():
    if df["SEQN"].isna().any() or df["SEQN"].duplicated().any():
        raise ValueError(f"{name} must have unique, nonmissing SEQN values.")

merged = files["DEMO_L"].copy()
for name in ["SLQ_L", "BMX_L", "DPQ_L"]:
    merged = merged.merge(files[name], on="SEQN", how="inner", validate="one_to_one")

structure_rows.append({
    "داده": "ادغام چهار فایل، پیش از محدودیت‌ها",
    "تعداد ردیف": len(merged),
    "تعداد افراد یکتا": merged["SEQN"].nunique(),
    "شناسه تکراری": int(merged["SEQN"].duplicated().sum()),
    "شناسه گمشده": int(merged["SEQN"].isna().sum())
})

structure_summary = pd.DataFrame(structure_rows)
structure_summary
```

## ۵.۲. کدهای ویژه و مقادیر نامعتبر

در آیتم‌های PHQ-9، کد ۷ یعنی امتناع و ۹ یعنی «نمی‌دانم»؛ هر دو گمشده می‌شوند. فقط پاسخ‌های ۰ تا ۳ معتبرند. در خواندن XPT، بعضی صفرها به شکل عددی بسیار کوچک دیده می‌شوند؛ فقط مقادیر نزدیک صفر با فاصله کمتر از ۱۰ به توان منفی ۱۲ به صفر برگردانده می‌شوند.

خواب باید یکی از مقادیر نیم‌ساعتی ۳ تا ۱۳٫۵ یا کدهای معتبر ۲ و ۱۴ باشد. دور کمر غیرمثبت یا مقدار غیرمتناهی نامعتبر است. مقادیر نامعتبر گمشده می‌شوند و در جدول جداگانه شمارش می‌شوند. برای دور کمر حد بالای دلخواه تعیین نمی‌شود.

منابع: [SLQ_L](https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/SLQ_L.htm)، [DPQ_L](https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/DPQ_L.htm) و [BMX_L](https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2021/DataFiles/BMX_L.htm).

</div>


```python
phq_items = [f"DPQ{i:03d}" for i in range(10, 100, 10)]
analysis_df = adults.copy()
phq_values = adults[phq_items].copy()
zero_artifacts = phq_values.notna() & phq_values.ne(0) & np.isclose(phq_values, 0, atol=1e-12, rtol=0)
phq_values = phq_values.mask(zero_artifacts, 0)
analysis_df[phq_items] = phq_values.where(phq_values.isin([0, 1, 2, 3]))
analysis_df["PHQ9_total"] = analysis_df[phq_items].sum(axis=1, min_count=9)

valid_sleep_values = [2, 14] + list(np.arange(3, 14, 0.5))
valid_masks = {col: adults[col].isin(valid_sleep_values) for col in ["SLD012", "SLD013"]}
valid_masks["BMXWAIST"] = np.isfinite(adults["BMXWAIST"]) & adults["BMXWAIST"].gt(0)
valid_masks["RIAGENDR"] = adults["RIAGENDR"].isin([1, 2])
for col, mask in valid_masks.items():
    analysis_df[col] = adults[col].where(mask)

audit_rows = []
for col in ["SLD012", "SLD013", "BMXWAIST", "RIAGENDR"] + phq_items:
    is_phq = col in phq_items
    invalid = phq_values[col].notna() & ~phq_values[col].isin([0, 1, 2, 3, 7, 9]) if is_phq else adults[col].notna() & ~valid_masks[col]
    audit_rows.append({
        "متغیر": col,
        "گمشده اولیه": int(adults[col].isna().sum()),
        "امتناع ۷": int(adults[col].eq(7).sum()) if is_phq else 0,
        "نمی‌دانم ۹": int(adults[col].eq(9).sum()) if is_phq else 0,
        "مقدار نامعتبر دیگر": int(invalid.sum()),
        "صفر اصلاح‌شده در خواندن فایل": int(zero_artifacts[col].sum()) if is_phq else 0,
        "گمشده پس از پاک‌سازی": int(analysis_df[col].isna().sum()),
        "درصد گمشدگی نهایی": analysis_df[col].isna().mean() * 100
    })
missing_audit = pd.DataFrame(audit_rows)
missing_audit
```

<div dir="rtl" lang="fa" style="text-align:right;line-height:1.9">

کدهای ۷۷۷۷۷ و ۹۹۹۹۹ در ساعت‌های خام خواب نیز به معنی امتناع و «نمی‌دانم» هستند. این ساعت‌ها متغیر اصلی تحلیل نیستند؛ کدهای ویژه آن‌ها برای استفاده احتمالی بعدی پاک می‌شوند. گمشدگی BMIWAIST به معنی گمشدگی دور کمر نیست؛ کد ۱ آن یعنی اندازه‌گیری انجام نشده است.

</div>


```python
time_columns = ["SLQ300", "SLQ310", "SLQ320", "SLQ330"]
time_rows = []
for col in time_columns:
    values = adults[col].astype("string").str.strip()
    refused = values.eq("77777").fillna(False)
    unknown = values.eq("99999").fillna(False)
    blank = values.eq("").fillna(False)
    analysis_df[col] = values.mask(refused | unknown | blank)
    time_rows.append({"متغیر": col, "امتناع ۷۷۷۷۷": int(refused.sum()), "نمی‌دانم ۹۹۹۹۹": int(unknown.sum()), "گمشده یا خالی اولیه": int((values.isna() | blank).sum())})
time_code_audit = pd.DataFrame(time_rows)
time_code_audit
```

## ۵.۳. ناسازگاری‌های منطقی

وجود دور کمر همراه با کد «اندازه‌گیری انجام نشده»، یا وجود داده معاینه برای فرد ثبت‌شده به‌عنوان «فقط مصاحبه»، ناسازگار است و پیش از ادامه باید بررسی شود. نمره PHQ-9 فقط با نه پاسخ معتبر ساخته می‌شود. گمشدگی ساعت خام برای خواب با کد ۲ یا ۱۴ به‌تنهایی ناسازگاری نیست، چون CDC این ساعت‌ها را برای حفظ محرمانگی حذف کرده است.

</div>


```python
logical_masks = {
    "دور کمر موجود همراه کد عدم اندازه‌گیری": adults["BMIWAIST"].eq(1) & adults["BMXWAIST"].notna(),
    "دور کمر موجود با وضعیت بدون معاینه بدنی": adults["BMDSTATS"].eq(4) & adults["BMXWAIST"].notna(),
    "دور کمر یا پاسخ افسردگی بدون وضعیت معاینه": (~adults["RIDSTATR"].eq(2)) & (adults["BMXWAIST"].notna() | adults[phq_items].notna().any(axis=1)),
    "نمره PHQ-9 خارج از دامنه": analysis_df["PHQ9_total"].notna() & ~analysis_df["PHQ9_total"].between(0, 27),
    "نمره PHQ-9 با آیتم ناقص": analysis_df["PHQ9_total"].notna() & analysis_df[phq_items].isna().any(axis=1)
}
logical_audit = pd.DataFrame({"بررسی": list(logical_masks), "تعداد ناسازگار": [int(mask.sum()) for mask in logical_masks.values()]})
if any(mask.any() for mask in logical_masks.values()):
    raise ValueError("Logical inconsistencies found. Review logical_masks before continuing.")
logical_audit
```

پس از آن، جنسیت و شرایط وزن و طرح بررسی می‌شوند. هر حذف فقط یک بار و نسبت به مرحله قبل شمرده می‌شود. فایل‌های ادغام‌شده پیش از پاک‌سازی در merged و بزرگسالان پاک‌سازی‌شده پیش از حذف در analysis_df باقی می‌مانند.

```python
weight_col = "WTMEC2YR"
design_columns = [weight_col, "SDMVSTRA", "SDMVPSU"]
valid_design = np.isfinite(analysis_df[design_columns]).all(axis=1) & analysis_df[weight_col].gt(1e-12)

flow_rows = [{"مرحله": "افراد مشترک در چهار فایل", "تعداد افراد": len(merged)}]
current = analysis_df.copy()
flow_rows.append({"مرحله": "سن معتبر ۲۰ سال و بالاتر", "تعداد افراد": len(current)})
for label, columns in [
    ("خواب روزهای کاری معتبر و موجود", ["SLD012"]),
    ("دور کمر معتبر و موجود", ["BMXWAIST"]),
    ("هر نه آیتم PHQ-9 معتبر و موجود", ["PHQ9_total"]),
    ("جنسیت معتبر و موجود", ["RIAGENDR"])
]:
    current = current.dropna(subset=columns).copy()
    flow_rows.append({"مرحله": label, "تعداد افراد": len(current)})
current = current.loc[valid_design.reindex(current.index)].copy()
flow_rows.append({"مرحله": "وزن مثبت و متغیرهای طرح کامل", "تعداد افراد": len(current)})

final_columns = ["SEQN", "RIDAGEYR", "RIAGENDR", "SLD012", "SLD013", "BMXWAIST", "PHQ9_total"] + phq_items + design_columns
final_df = current[final_columns].copy().reset_index(drop=True)
sample_flow = pd.DataFrame(flow_rows)
sample_flow["حذف‌شده نسبت به مرحله قبل"] = sample_flow["تعداد افراد"].shift().sub(sample_flow["تعداد افراد"]).fillna(0).astype(int)
sample_flow
```

```python
def fa_number(value, digits=1):
    return f"{value:,.{digits}f}".translate(str.maketrans("0123456789,.", "۰۱۲۳۴۵۶۷۸۹٬٫"))

summary_indexed = quantitative_summary.set_index("متغیر")
pattern_rows = []
for row in sex_frequency.to_dict("records"):
    pattern_rows.append({"موضوع": "ترکیب نمونه", "توضیح": f"{row['جنسیت ثبت‌شده']}: {fa_number(row['فراوانی'], 0)} نفر، معادل {fa_number(row['درصد'])} درصد نمونه."})

for label in ["سن", "خواب روزهای کاری", "دور کمر", "نمره PHQ-9"]:
    row = summary_indexed.loc[label]
    pattern_rows.append({"موضوع": label, "توضیح": f"میانه {fa_number(row['میانه'])} و نیمی از مقادیر بین {fa_number(row['چارک اول'])} و {fa_number(row['چارک سوم'])} {row['واحد']} است."})

phq_mean = final_df["PHQ9_total"].mean()
phq_median = final_df["PHQ9_total"].median()
phq_skew = final_df["PHQ9_total"].skew()
shape = "به سمت راست کشیده است" if phq_skew > 0 else "به سمت چپ کشیده است" if phq_skew < 0 else "چولگی ندارد"
pattern_rows.append({"موضوع": "شکل توزیع PHQ-9", "توضیح": f"میانگین {fa_number(phq_mean)} و میانه {fa_number(phq_median)} است؛ ضریب چولگی {fa_number(phq_skew, 2)} است و توزیع {shape}."})
zero_count = int(final_df["PHQ9_total"].eq(0).sum())
pattern_rows.append({"موضوع": "نمره صفر PHQ-9", "توضیح": f"{fa_number(zero_count, 0)} نفر، معادل {fa_number(zero_count / len(final_df) * 100)} درصد، نمره صفر دارند."})

for col, value, label in [("RIDAGEYR", 80, "سن ۸۰ سال و بالاتر"), ("SLD012", 2, "خواب کمتر از ۳ ساعت"), ("SLD012", 14, "خواب ۱۴ ساعت یا بیشتر")]:
    number = int(final_df[col].eq(value).sum())
    pattern_rows.append({"موضوع": label, "توضیح": f"{fa_number(number, 0)} نفر، معادل {fa_number(number / len(final_df) * 100)} درصد نمونه؛ مقدار ثبت‌شده زمان یا سن دقیق این افراد نیست."})
important_patterns = pd.DataFrame(pattern_rows)
important_patterns
```

چولگی مثبت یعنی دنباله توزیع به سمت مقادیر بالا و چولگی منفی یعنی دنباله به سمت مقادیر پایین کشیده است. این شواهد برای انتخاب خلاصه‌ها به کار می‌روند و به‌تنهایی ثابت نمی‌کنند که توزیع نرمال است.

```python
def fa_value(value):
    return f"{value:.2f}".translate(str.maketrans("0123456789.-", "۰۱۲۳۴۵۶۷۸۹٫−"))

evidence = distribution_evidence.set_index("متغیر")
interpretation_rows = []
reasons = {
    "سن": "چولگی منفی خفیف است؛ میانگین و انحراف معیار خلاصه کلی مفیدی هستند، اما میانه و چارک‌ها نیز لازم‌اند چون میانگین پایین‌تر از میانه است. ثبت سن ۸۰ سال و بالاتر به صورت ۸۰، پراکندگی واقعی سن را محدود می‌کند.",
    "خواب روزهای کاری": "میانگین به میانه نزدیک و چولگی کم است؛ بنابراین میانگین و انحراف معیار برای خلاصه کلی مناسب‌اند. میانه و دامنه بین چارکی نیز نیمه میانی خواب را نشان می‌دهند. کدهای ۲ و ۱۴ همچنان زمان دقیق نیستند.",
    "دور کمر": "میانگین کمی بیشتر از میانه و چولگی مثبت است؛ دور کمرهای بالا میانگین و انحراف معیار را بیشتر تحت تاثیر قرار می‌دهند. این دو معیار مفیدند، اما بهتر است همراه میانه و دامنه بین چارکی گزارش شوند.",
    "نمره PHQ-9": "میانگین از میانه بالاتر و چولگی راست بیشتر از سایر متغیرهاست. چارک اول صفر است و نمره‌های بالا میانگین را بالا می‌برند؛ بنابراین میانه و دامنه بین چارکی برای توصیف مرکز و پراکندگی نیمه میانی مناسب‌ترند. میانگین و انحراف معیار به‌تنهایی کافی نیستند."
}
for label in reasons:
    row = quantitative_summary.set_index("متغیر").loc[label]
    skew = evidence.loc[label, "چولگی نمونه"]
    interpretation_rows.append({
        "متغیر": label,
        "شواهد نمونه": f"میانگین {fa_value(row['میانگین'])}، میانه {fa_value(row['میانه'])}، انحراف معیار {fa_value(row['انحراف معیار'])}؛ چارک‌ها {fa_value(row['چارک اول'])} و {fa_value(row['چارک سوم'])}، دامنه بین چارکی {fa_value(row['دامنه بین چارکی'])}؛ چولگی {fa_value(skew)}.",
        "دلیل انتخاب معیار": reasons[label]
    })
measure_interpretation = pd.DataFrame(interpretation_rows)
measure_interpretation
```

## الف۳. مقایسه خلاصه خام و وزن‌دار

برای هر دو روش، از همان افراد final_df استفاده می‌شود. در میانگین خام، سهم همه افراد برابر است. در میانگین وزن‌دار، سهم هر فرد بر اساس وزن معاینه WTMEC2YR تعیین می‌شود: مجموع حاصل‌ضرب مقدار متغیر در وزن، تقسیم بر مجموع وزن‌ها.

ستون تفاوت برابر میانگین وزن‌دار منهای میانگین خام است و واحد آن با متغیر یکسان است.

</div>


```python
weights = final_df["WTMEC2YR"]
if weights.isna().any() or not np.isfinite(weights).all() or weights.le(0).any():
    raise ValueError("Weights must be finite, positive, and nonmissing.")

comparison_rows = []
for col, label, unit in [
    ("SLD012", "خواب روزهای کاری", "ساعت"),
    ("BMXWAIST", "دور کمر", "سانتی‌متر"),
    ("PHQ9_total", "نمره PHQ-9", "امتیاز")
]:
    raw_mean = final_df[col].mean()
    weighted_mean = np.average(final_df[col], weights=weights)
    comparison_rows.append({
        "متغیر": label,
        "واحد": unit,
        "تعداد افراد": len(final_df),
        "میانگین خام": raw_mean,
        "میانگین وزن‌دار": weighted_mean,
        "تفاوت وزن‌دار منهای خام": weighted_mean - raw_mean
    })
weighted_comparison = pd.DataFrame(comparison_rows)
weighted_comparison
```

```python
comparison_notes = []
for row in weighted_comparison.to_dict("records"):
    difference = row["تفاوت وزن‌دار منهای خام"]
    direction = "بیشتر" if difference > 0 else "کمتر" if difference < 0 else "برابر"
    comparison_notes.append({
        "متغیر": row["متغیر"],
        "توضیح": f"میانگین خام {fa_value(row['میانگین خام'])} و میانگین وزن‌دار {fa_value(row['میانگین وزن‌دار'])} است؛ مقدار وزن‌دار {fa_value(abs(difference))} {row['واحد']} {direction} است."
    })
weighted_interpretation = pd.DataFrame(comparison_notes)
weighted_interpretation
```

میانگین‌های خام و وزن‌دار متفاوت‌اند، اما بسیار به هم نزدیک هستند. در این نمونه، وزن‌دهی تغییر کمی در میانگین خواب، دور کمر و PHQ-9 ایجاد کرده است. افراد نماینده تعدادهای متفاوتی از جمعیت هستند؛ وزن‌ها احتمال انتخاب متفاوت، عدم پاسخ کلی و تطبیق با تعدادهای جمعیتی را در نظر می‌گیرند. جهت و اندازه تفاوت به مقادیر متغیر در افراد دارای وزن بیشتر یا کمتر وابسته است.
